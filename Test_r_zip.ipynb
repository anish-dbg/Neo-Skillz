Zip function
- return zip object 
- iterates pair

```python
l1 = [1,2,3,4]
l2 = [5,6,7,8]
list(zip(l1,l2))
```

```python
# roll no and name from database
roll_no = [1,2,3,4,5]
student_name = ['Bappy','Anish','Nitish','Vikash','Riyan']
list(zip(roll_no,student_name))
```

```python
[i*j for i,j in zip(l1,l2)]
```

```python
l3 = [9,10,11,12]
list(zip(l1,l2,l3))
```

```python
for i in zip(l1,l2):
    print(i)
```
