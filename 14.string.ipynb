String:
- It's a Sequence of characters
- In python specifically, strings are a sequence of unicode characters

Topic to be discussed
- Creating strings
- Accessing strings
- Adding chars to strings
- editing strings
- Deleting strings
- Operations on strings
- string functions

```python
# Creating strings
s = 'hello'
s = "hello"
# multiline strings
s = '''hello'''
s = """hello"""
s = str('hello')
print(s)
```

```python
# Accessing strings
s = "welcome to python"
print(s[11])     # positive indexing left to right
print(s[-1])     # negative indexing right to left
```

```python
# string slicing  -- accesing multiple elements
print(s[0:7])  # last number is always excluded
```

```python
print(s[0:])
```

```python
print(s[:])
```

```python
print(s[0:7:2]) # jump parameter (a-1) ignore it
```

```python
print(s[7:0:-1]) # reverse 
```

```python
s = "welcome to python"
print(s[::-1])
```

```python
print(s[-6:])
```

```python
print(s[-1:-7:-1])
```

```python
s[0] = 'A'   # python strings are immutable
```

```python
del s[0]   # we can't delete 
```

String operations
- Arithmetic Operations
- Relational Operations
- Logical Operations
- Loops on Strings
- Membership Operations

```python
# Arithmetic operations
print("Hi" +" "+ "Everyone")   # concatination
```

```python
print("Hi "*10)
```

```python
# Relational Operations   ( ascii code  value compare)

print("Hello" == "Hello")
print("Hello" != "World")
print("Hello" > "Hi")
print("Hello" < "World")
print("Hello" >= "Hello")
```

```python
# logical operations
"hello" and "world"
```

```python
"hello" or "world"
```

```python
"hello" or ""
```

```python
"hello" and ""
```

```python
not "hello"
```

```python
not ""
```

```python
# loops on strings
for i in "Bappy":
    print(i)
```

```python
for i in "bappy":
    print("hello")
```

```python
# membership operator
'B' in "Bappy"
```

```python
'B' not in "Bappy"
```
