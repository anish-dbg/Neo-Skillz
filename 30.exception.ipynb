Exceptions

If things go wrong during the execution of the program(runtime). it generally happens when something unforseen has happend.

- Exceptions are raised by python runtime
- you have to take is on the fly

Examples
- Memory overflow
- Divide by 0 -> logical error
- Database error

```python
def divide(num):
    r = 12/num
    print(r)

divide(0)
```

Why is it important to handle exceptions?
- catch and handle the error instead of crashing the whole systems

How to handle exceptions -> Try except block

```python
# let's create a file
with open('files/sample.txt','w') as f:
    f.write('hello world')
```

```python
# let's read a file
try:
    with open('files/sample.txt','r') as f:
        print(f.read())
except:
    print("file not found")
```
