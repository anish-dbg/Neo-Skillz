Logging in python

```python
import logging
import os

log_dir = "logs"
log_file_name = "application.log"

os.makedirs(log_dir, exist_ok=True)
log_path = os.path.join(log_dir, log_file_name)

logging.basicConfig(
    filename=log_path,
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s",
    force=True,   # <-- important
)

logging.info("How are you?")
```
