 Literals:
 - Literals is a fixed values used to identify variables, function, classes and all other elements used in programming language

```python
a = 0b1010 # Binary Literals
b = 100    # Decimal Literals
c = 0o310  # Octal Literals
d = 0x12c  # Hexadecimal Literals
```

```python
print(a,b,c,d)
```

```python
# Float literals

float_1 = 10.5
float_2 = 1.5e2   # 1.5*10^2
float_3 = 1.5e-3  # 1.5*10^-3
```

```python
print(float_1)
print(float_2)
print(float_3)
```

```python
# Complex Literals
x = 3.4j
```

```python
type(x)
```

```python
print(x.real)
print(x.imag)
```

```python
string = 'This is a Python'
strings = "This is a Python"
char = 'C'
multiline_str = """This is a multiline string with more than one line code"""
unicode = u"\U0001F600\U0001F604\U0001F602"
raw_str = r"raw \n string"  # it ignore special escape character

print(string)
print(strings)
print(multiline_str)
print(char)
print(unicode)
print(raw_str)
```

```python
# True = 1
# False = 0

a = True + 4
b = False + 10

print("a:",a)
print("b:",b)
```
