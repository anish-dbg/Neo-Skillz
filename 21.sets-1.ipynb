```python
# len/sum/min/max/sorted

s = {3,1,4,5,2,7}

print(len(s))
print(sum(s))
print(min(s))
print(max(s))
print(sorted(s,reverse=True))
```

```python
# union/update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

# s1 | s2
s1.union(s2)
```

```python
s1.update(s2)
print(s1)
print(s2)
```

```python
# intersection/ intersection_update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

s1.intersection(s2)
print(s1)
print(s2)
```

```python
s1.intersection_update(s2)
print(s1)
print(s2)
```

```python
# difference/difference_update
s1.difference(s2)
```

```python
s1.difference_update(s2)
print(s1)
print(s2)
```

```python
# symmentric_difference/symmentic-difference_update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

s1.symmetric_difference(s2)
```

```python
s1.symmetric_difference_update(s2)
print(s1)
print(s2)
```

```python
# isdisjoint ka matlab hota hai ki koi common value nhi hona chahiye
s1 = {1,2,3,4}
s2 = {7,8,5,6}

s1.isdisjoint(s2)
```

```python
# subset

s1 = {1,2,3,4,5}
s2 = {3,4,5}

s2.issubset(s1)
```

```python
# superset
s1 = {1,2,3,4,5}
s2 = {3,4,5}

s1.issuperset(s2)
```

```python
# copy
s1 = {1,2,3}
s2 = s1.copy()
print(s2)
```

## Frozenset

frozen set is just an immutable version of a python set object

```python
fs1 = frozenset([1,2,3,4])
fs2 = frozenset([3,4,5,6])
print(fs1)
print(fs2)
print(type(fs1))
```

```python
fs1 | fs2
```

```python
fs1 & fs2
```

```python
fs1 - fs2
```

```python
fs1 ^ fs2
```
