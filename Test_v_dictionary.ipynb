## Dictionary

Dictionary in python is a collection of keys values, used to store data values likes a map, which unlike other data types which hold only a single value as an element.
in some languages it is known as map or associative arrays.

dict = { 'name': 'anish', 'age': 22, 'gender': 'male'}

Characteristics:
- Mutable
- Indexing has no meaning
- keys can't be duplicated
- keys can't be mutable items

```python
# empty dictionary
d = {}
print(d)
print(type(d))
```

```python
# 1D dictionary
d1 = { 'name': 'anish', 'age': 22, 'gender': 'male'} # double quotes --> single quotes
print(d1)
```

```python
# with mixed keys
d2 = {(1,2,3):33, "name": "Anish"}   # immutable 
print(d2)
```

```python
# 2d dictionary
d3 = {
    "name": "Anish",
    "college_name": "LNCT",
    "semester": 5,
    "subjects": {"DSA": 90, "MATH": 96, "EnglISH": 85}
}

print(d3)
```

```python
# using sequence and dict function

d4 = dict([("name","Anish"), ("age", 22)])
print(d4)
```

```python
# duplicate keys
d5 = { "name": "Anish","age": 22, "gender": "male", "name": "Nitish"}
print(d5)  # always take latest one
```

```python
# Mutable items as keys

d6 = {[1,2,3]:33, "name": "Bappy"} # dictionary mai mutable items nhi hone chahiyae
print(d6)
```

```python
# Accessing items from  the dict
d1 = {"name": "Anish", "age": 22, "Gender": "Male"}
d1["name"]
d1["age"]
```

```python
d3 = {
    "name": "Anish",
    "college_name": "LNCT",
    "semester": 5,
    "subjects": {"DSA": 90, "MATH": 96, "ENGLISH": 85}
}

d3["subjects"]["DSA"]
d3["subjects"]["ENGLISH"]
```

```python
# adding new key-value pair
d3["weight"] = 70
```

```python
print(d3)
```

```python
d3["subjects"]["python"] = 89
```

```python
print(d3)
```

```python
# edit the existing keys
d3["name"] = "Riyan Venkatesh"
print(d3)  # value we can change but key we can't change
```

```python
# Remove key-value pair
d3 = {
    "name": "Anish",
    "age": 22,
    "college_name": "LNCTU",
    "semester": 5,
    "subjects": {"DSA": 90, "MATH": 96, "ENGlISH": 85}
}
d3.pop("age")
print(d3)
```

```python
# pop item
d3.popitem()  # remove the last item
print(d3)
```

```python
d3 = {
    "name": "Anish",
    "age": 22,
    "college_name": "LNCTU",
    "semester": 5,
    "subjects": {"DSA": 90, "MATH": 96, "ENGlISH": 85}
}

del d3["name"]
print(d3)
```

# Dictionary Operations
- Membership
- Iteration

```python
# works on key
d = {"name": "Anish", "age": 22,"gender": "male"}
"name" in d
```

```python
# iteration

for i in d:
    print(i,":",d[i])
```

Dictionary Methods and functions

```python
d = {"name": "Bappy", "age": 22, "gender": "male"}
print(d)
```

```python
# Common functions
print(len(d))
print(sorted(d,reverse=True)) # decending order
print(max(d))  # ascii value
print(min(d))
```

```python
# items/keys/values
d = {"name": "Anish", "age": 22, "gender": "male"}
print(d.items())
```

```python
print(d.keys())
```

```python
print(d.values())
```

```python
# update 
d1 = {1:2,3:4,4:5}
d2 = {4:7,6:8}

d1.update(d2)
print(d1)
```

Dictionary Comprehension
- {key: value for vars in iterable}

```python
# print 1st 10 numbers and their squares
# 1**2 --> square root

{ i: i*i for i in  range(1,11)}
```

```python
products = {"Phone": 10, "laptop": 0, "Charger": 32, "tablet": 0}
for key,val in products.items():
    print(key,val)
```

```python
{key:value for (key,value) in products.items() if value>0}
```

```python
# using zip

days = ["Sunday", "Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday"]
temp_c = [30.5,32.6,31.8,33.4,29.8,30.2,29.9]

{i:j for (i,j) in zip(days,temp_c)}
```
