What are Lists
- List is a data type where you can store multiple items under 1 name.
- More technically, lists act like dynamic arrays which means you can add more items on the fly


Array
- fixed size
- homogeneous type
- speed of execution fast
- less memory

Lists
- Dynamic size
- Hetrogeneous type
- speed of execution is slow
- more memory

```python
# Why Lists are required in programming?
L = [1,2,3]

print(id(L))
print(id(L[0]))
print(id(L[1]))
print(id(L[2]))
```

```python
print(id(1))
print(id(2))
print(id(3))
```

Lists Methods in python

```python
#append
l = [1,2,3,4]
print(l)
l.append(5)
print(l)
```

```python
l[0] = 'anish'
```

```python
print(l)
```

```python
l = [1,2,3,4]
l[1:4] = [100,200,300]
```

```python
print(l)
```

```python
l = [1,2,3,4]
l.append([6,7,8])
print(l)
```

```python
# extend
l = [1,2,3,4]
a = [10,11,12]
l.extend(a)
print(l)
```

```python
l.extend("Hello")
print(l)
```

```python
# insert
l = [1,2,3,4]
l.insert(1,200)  # (place,value)
print(l)
```

```python
# Deleting items from a list
l =  [ 1,2,3,4,5]
del l[0]
del l[1:3]
print(l)
```

```python
# remove
l = [1,2,3,4,5]
l.remove(4)  # remove(value)
print(l)
```

```python
print(l)
```

```python
# pop
l = [1,2,3,4]
l.pop()   # pop(index) ke basis by default last element remove
print(l)
```

```python
print(l)
```

```python
# Clear
l = [1,2,3,4]
l.clear()
print(l)
```

```python
# len/min/max/sorted

l = [1,2,3,4,5]
print(len(l))
print(min(l))
print(max(l))
print(sorted(l,reverse=True))
```

```python
# Count
l = [1,2,3,3,4,4,4,5,6]
l.count(4)
```

```python
# index
l.index(5)
```

```python
# reverse
l = [1,2,3,4]
l.reverse()
print(l)
```

```python
# sort
l = [3,4,1,3,5,1,2]
l.sort()
print(l)
```

```python
# Copy
l = [1,2,3,4]
l1 = l.copy()
print(l1)
```
