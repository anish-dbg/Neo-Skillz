Variable
- variable is a container which is used to store any kind of value

why?
- A variable is used to store data temporarily in memory so that we can use, modify, and reuse it later in a program.

```python
a = 10   # a = variable and 10 is value
b = 10.1
c = "Hello"
d = True
print(a)
print(b)
print(c)
print(d)
```

```python
# Dynamic Binding: means in which we further initialize same variable name to different data types and it is resolved in runtime time
# Dynamic binding means the method call is resolved at runtime based on the object.

# Static Binding: means in which we can't further initialize same variable name to different data type and it is resolved in compile time
# The method/function call is decided at compile time.

# Compile time
# Compile time is the phase when the source code is checked and translated into machine code before execution.

# Runtime
# Runtime is the phase when the program is running and executing instructions.
```

```python
a = 10
a = 'anish'
print(a)
```

```python
# python is dynamic typed language
```

```python
a,b,c = 1,2,3
print(a,b,c)
```

```python
a = 5
b = 6
c = 7
print(a,b,c)
```

```python
a=b=c = 5
print(a,b,c)
```
