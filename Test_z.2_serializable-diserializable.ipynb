Serialization and Deserialization
- Serialization - process of converting python data types to json format
- Deserailization - process of converting json to python data types

what is json?
- The full form of json is javascript object notation
- file format
- it supports all programming language

```python
# serialization using json module
# list

import json

L = [1,2,3,4,5,6,7,8]

with open("files/test.json",'w') as f:
    json.dump(L,f)
```

```python
# dict

d = {
    'name': 'Anish',
    'age': 22,
    'gender': 'male'
}

with open('files/test.json','w') as f:
    json.dump(d,f, indent=4)
```

```python
# deserialiazation

import json

with open('files/test.json','r') as f:
    data = json.load(f)
    print(data)
    print(type(data))
```
