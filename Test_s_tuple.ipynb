## Tuples
A tuples in python is similar to a list. The difference between the two is that we cannot change the elements of a tuple once it is assigned whereas we can change the elements of a list.

Characterstics
 - Ordered
 - Unchangeble
 - Allows duplicate

Topic to be discussed
 - Creating a Tuple
 - Accessing items
 - Editing items
 - Adding items
 - Deleting items
 - Operations on Tuples
 - Tuple Functions

```python
# Creating Tuple
t1 = () # parenthesis
print(t1)
print(type(t1))
```

```python
# create a tuple with a single item
t2 = ("Bappy")
print(t2)
print(type(t2))
```

```python
t2 = ("Bappy", )
print(t2)
print(type(t2))
```

```python
# homo
t3 = (1,2,3,4)
print(t3)

# hetro
t4 = (1,2.5,True,[1,2,3])
print(t4)

# nested tuple
t5 = (1,2,3,(4,5))
print(t5)

# using type conversion
t6 = tuple('hello')
print(t6)
```

```python
# Accessing elements
 # indexing
 # slicing

t3[0]
```

```python
t3[-2]  # from back-side
```

```python
# slicing
t3[0:3]
```

```python
t3[0:3:2]   # (a-1)
```

```python
t3[::2]
```

```python
# Editing and deleting elements
t3 = (1,2,3,4)
t3[0] = 100
```

```python
del t3[0]
```

```python
t3 = (1,2,3,4)
temp = list(t3)
print(temp)
temp.append(200)
temp
```

```python
t3 = tuple(temp)
t3
```

## Tuples Operations in python
 - Arithmetic
 - Membership 
 - loop

```python
t1 = (1,2,3,4)
t2 = (5,6,7,8)

t1 + t2  # concatination
```

```python
t1*4
```

```python
t1-t2
```

```python
# membership 
t1 = (1,2,3,4)
1 in t1
```

```python
1 not in t1
```

```python
# loop
t1 = (1,2,3,4)

for i in t1:
    print(i)
```

```python
# tuples methods in python
t = (1,2,3,40)

print(len(t))
print(min(t))
print(max(t))
print(sum(t))
print(sorted(t,reverse=True))
```

```python
t.count(3)
```

```python
t.index(3)
```

## Difference between Lists and Tuples

 - Syntax
 - Mutability
 - Speed
 - Memory
 - Built in Functionality
 - Error prone
 - usability

- [] - lists
- () - tuples
- lists - mutable
- tuples - immutable
- lists - slow (mutable)
- tuples - fast (immutable)
- tuples - less space
- lists - more space , more built-in functionality

```python
import sys

L = list(range(1000))
T = tuple(range(1000))

print('List size', sys.getsizeof(L))
print('Tuple size', sys.getsizeof(T))
```

```python
a = [1,2,3]
b = a

a.append(4)
print(a)
print(b)
```

```python
a = (1,2,3)
b = a
a = a + (4,)
print(a)
print(b)  # copy
```

```python
# tuple unpacking

a,b,c = (1,2,3)
print(a,b,c)
```

```python
a,b,c =[1,2,3]
print(a,b,c)
```

```python

# zipping tuples
a = (1,2,3,4)
b = (5,6,7,8)

tuple(zip(a,b)) # pair
```

```python
# Tuple Comprehension

TC = tuple((i for i in range(1,11)))
print(TC)
```

```python
# Scaler multiplication on a vector
v = [2,3,4]
s = -3

L = []

for i in v:
    result = i*s
    L.append(result)
print(L)
```

```python
res = tuple((i*s for i in v))
print(res)
```
