Data Types:
- A data type is an attribute attached to a piece of data that tells a computer system how to interpret and use its value.

Why data types matter?
- memory allocation
- Operation Rules
- System Efficiency

Data Types
- Integer(int)
- Float(float/decimal)
- Boolean(bool- true/false)
- Complex
- String(str)

```python
# Integer : -3,-2,-1,0,1,2,3
print(-1)
print(0)
print(1)
```

```python
print(int(10))  # int() - is a function which converts into integer
print(int(10.1))
# print(int(2+4j))  error
print(int(True))
print(int(False))
# print(int("hello"))   value error
print(int("10"))
```

```python
# float : -1.0,2.5
print(-1.0)
print(0)
print(2.5)
```

```python
print(float(10))  # float() is a function which converts into floating numbers
print(float(10.1))
# print(float(2+4j))  error
print(float(True))
print(float(False))
# print(float("hello"))  error
print(float("10"))
```

```python
# type() - a function used to detect the type of data
type(2)
```

```python
type(2.5)
```

```python
# Boolean 
# python is a case sensitive language and interpreted language
print(True)
```

```python
print(true)
```

```python
print(bool(0)) # bool() - is a function which converts into boolean
print(bool(1))
print(bool(10.1))
print(bool(10))
print(bool(0+2j))
print(bool(0+0j))
print(bool("anish"))
print(bool("False"))
print(bool(False))
```

```python
# String : A Collection of symbols or characters
print("Hello")
type("Hello")
```

```python
print(str(10)) # str() is a function that converts into string
print(str(10.2))
print(str(1+3j))
print(str(True))
```

```python
# Complex: It contain real number + Imaginary number
print(2+3j)
type(2+3j)
```

```python
print(complex(10)) # complex() is a function that converts into complex
print(complex(12.3))
print(complex(False))
print(complex("12"))
print(complex(12,3))
```
