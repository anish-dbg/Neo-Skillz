```python
import time
print('Hello without flush', end=' ') # Output is buffered
time.sleep(1)
print('World!') # Output might appear all at once after the delay

print('Hello with flush', end=' ', flush=True) # Output is forced immediately
time.sleep(1)
print('World!') # Output appears after the 'Hello with flush' has already shown up
```
