File handling in python

Types of data used for i/o
- Text - '1234' as a sequence of unicode
- Binary - 1234 as a sequence of bytes of its binary equivalent

Hence there are 2 files types to deal with
- Text files - all program files are text files
- Binary files - Images, music,video,exe files

How File i/o is done in most programming languages
- Open a file
- Read/write data
- Close the file

writing to a file

```python
# if the file is not present
f = open("files/sample.txt","w") # it returns object
f.write("Hello Python")
f.close()

# f.write("Hello")
```

```python
# write multiple strings
f = open('files/sample1.txt','w')
f.write('Hello World')
f.write('\nhow are you?')
f.close()
```

```python
# case 2 - if the file is already present
f = open('files/sample.txt','w')  # older file ko replace krna hoga to w(write mode)
f.write('yo yo honey singh')
f.close()
```

```python
# introducing append mode
f = open('files/sample.txt','a')
f.write('\n I am fine')
f.close()
```

```python
# write lines
L = ['hello\n','hi\n','how are you\n', 'I am fine']

f = open('files/sample.txt','w')
f.writelines(L)
f.close()
```

Reading from files

```python
f = open('files/sample.txt','r')
s = f.read()
print(s)
f.close
```

```python
# reading upto n chars

f = open('files/sample.txt','r')
s = f.read(10)
print(s)
f.close()
```

```python
# readline() -> to read line by line
f = open('files/sample.txt','r')
print(f.readline(),end='') # by default print give \n
f.close()
```

```python
# reading entire using readline
f = open('files/sample.txt','r')

while True:
    data = f.readline()

    if data == "":
        break
    else:
        print(data,end='')
```

File Handling using Context manager (with)

Using Context manager(with)
- it's a good idea to close a file after usage as it will free up the resources
- if we don't close it, garbage collector would close it
- with keyword closes the files as soon as the usage is over

```python
with open('files/sample.txt','w') as f:
    f.write("Hello there!")
```

```python
# read()
with open('files/sample.txt','r') as f:
    s = f.read() # f is an instance of sample.txt
    print(s)
```

```python
# readline()

with open('files/sample.txt','r') as f:
    s = f.readline()
    print(s)
```

```python
# append mode
with open('files/sample.txt', 'a') as f:
    s = f.write("\nMy Name is Anish")
```

```python
with open("files/sample.txt",'r') as f:
    print(f.read(10))
    print(f.read(10))
```

```python
# benefit? -> to load a big file in memory

big_L = ['hello world\n' for i in range(1000)]

with open('files/big.txt','w') as f:
    f.writelines(big_L)
```

```python
with open('files/big.txt','r') as f:
    chunk_size = 10

    while len(f.read(chunk_size)) > 0:
        print(f.read(chunk_size), end="")
        f.read(chunk_size)
```

```python
# seek and tell function

with open('files/sample.txt','r') as f:
    print(f.read(10))
    print(f.read(10))
    print(f.tell()) # cursor pointer
```

```python
# seek

with open('files/sample.txt','r') as f:
    print(f.read(11))
    f.seek(0) # set the cursor pointer at the new position

    print(f.tell())
```

```python
with open("files/sample.txt",'w') as f:
    f.write("Hello")
    f.seek(0)
    f.write("D")
    print(f.tell())
```

Problems with working in text mode
- can't work with binary files like images
- not good for other data types like int/float/list/tuples

```python
with open("files/icon.jpeg",'r') as f:
    print(f.read())
```

```python
with open("files/icon.jpeg",'rb') as f:
    print(f.read())
```

```python
# working with binary files
with open("files/icon.jpeg",'rb') as f:
    with open("files/icon_copy.png",'wb') as wf:
        wf.write(f.read())
     
```

```python
# working with other data types
with open('files/sample.txt','w') as f:
    f.write("5")
```

```python
# more complex data

d = {
    'name': 'anish',
    'age': 22,
    'gender': 'male'
}

with open('files/sample.txt','w') as f:
    f.write(str(d))
```

```python
with open('files/sample.txt','r') as f:
    s = f.read()
    print(s)
    print(type(s))
```

```python
with open('files/sample.txt','r') as f:
    s = f.read()
    print(s)
    print(dict(s)) # not exits
```
