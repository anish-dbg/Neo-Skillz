Operators:
- To perform operation between two or more than   
two operands, a symbol is used to perform a particular operation that is known as operators

Types of Operators:
- Arithmetic Operators
- Relational Operators
- Logical Operators
- Bitwise Operators
- Assignment Operators
- Membership Operators
- Ternary Operators

```python
# Arithmetic Operators
a = 2+3
b = 4-5
c = 4*5
d = 4/2  # division operator it returns float value
e = 4//2  # integer division operator it return integer value
f = 4%2   # modulus operator it return remainder as integer value 

print(a)
print(b)
print(c)
print(d)
print(e)
print(f)
```

```python
# interesting things
print(2.45%2)
print(5/2)
print(5.0/2)
print(5.0/2.0)
print(4/2.0)
```

```python
# Relational operator : it return boolean value
print(4>5)
print(4<5)
print(4>=5)
print(4<=5)
print(4==4)
print(4!=4)
```

```python
# Logical Operator
print( 1 and 0 )
print( 1 and 1 )
print( 0 and 0 )
print( 0 and 1 )
print( 0 or 0 )
print( 0 or 1 )
print( 1 or 1 )
print( 1 or 0 )
```

```python
print( not 1)
print( not 0)
```

```python
# Bitwise Operator

print(2 & 3) # and operator
print(2 | 3) # or operator
print(2 ^ 3) # exor operator ( same same = 0 and diff diff = 1)
print (~2)   # reverse ( -(a+1)) first find 2's complement store and knowing the number then add 1 to original binary number to the last
print( 2>>3) # right shift ( ( a/2^b)
print( 2<<3) # left shift ( (a*2^b)
```

```python
# Assignment Operator

a = 2 # 2 is assigned to a 
b += 2
c -= 2
print(a)
print(b)
print(c)
# ++a ( first add then use) , a++ ( first use then add), --a, a---
```

```python
# Membership operator (in) / (not in) is used to identify whether the elements present or not in expression
print('m' in 'mumbai')
print('m' not in 'mumbai')

# it is robust implementation
'''
problems without crashing
Handles errors properly
Does not fail easily
Works correctly even with unexpected input
'''
```
