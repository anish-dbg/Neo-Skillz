loop-conditional;
- Break (break the loop and comes out)
- Continue ( skip )
- pass  ( don't know what to do at that time)

```python
# Break

for i in range(1,11):
    if i == 5:
        break
    print(i)
```

```python
name = ["Alex", "John","Bappy", "Mr. X","Mr. Y"]

for i in name:
    if i == "Bappy":
        print(i)
        break
```

```python
# find prime numbers in a range

for i in range(1,11):
    if (i%2 != 0):
        print(i," ")
```

```python
for i in range(2,11):
    for j in range(2, i):
        if i%j == 0:
            break
    else:
        print(i)
```

```python
# continue

for i in range(1,10):
    if i == 5:
        continue
    print(i)
```

```python
# pass

for i in range(1,10):
    pass
```
