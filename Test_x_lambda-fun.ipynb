Lambda functions in python
- A lambda function is a small anonymous function ( a function defined without name)
- it can take any number of arguments, but can only have one expressions
- it is an one line function

lambda a,b: a+b

```python
a = lambda x: x**2
a(4)
```

```python
a = lambda x,y: x+y
a(4,6)
```

Diff between lambda vs normal function

- No name
- lambda has no return value(infact, returns a function)
- lambda is written in 1 line
- not reusable

Then why use lambda functions?
- They are used with higher order function(hof)

```python
# check if a string has 'a'
a = lambda s: 'a' in s
a('hello')
```

```python
# odd or even

a = lambda x:'even' if x%2 == 0 else 'odd'
a(6)
```

Higher order function
- those function who takes input as a function
- when function returns a function

```python
l = [1,2,3,4,5]

def square(x):
    return x**2

def cube(x):
    return x**3

def transform(f,l):  # hof
    output = []
    for i in l:
        output.append(f(i))
    print(output)
```

```python
# transform(square,l)
transform(cube,l)
```

```python
transform(lambda x:x**2, l)
```

Higher Order Function
- map
- filter
- reduce

map

```python
# square the items of a list
list(map(lambda x:x**2, [1,2,3,4,5]))
```

```python
a = lambda x:x
l = [1,2,3,4,5]

map(a,l)
```

```python
# odd/even labelling of list items
l =[1,2,3,4,5]
list(map(lambda x:'even' if x%2 == 0 else 'odd', l ))
```

```python
# fetch names from a list of dict

users = [
    {
        'name': 'Alex',
        'age': 30,
        'gender': 'male'
    },
    {
        'name': 'Bappy',
        'age': 23,
        'gender': 'male'
    },
    {
        'name': 'Anish',
        'age': 22,
        'gender': 'male'
    }
]

list(map(lambda users:users['name'],users))
```

filter

```python
# numbers greater than 5
l = [1,2,3,4,5]
list(filter(lambda x:x>2,l))
```

```python
# fetch fruits starting with 'a'

fruits = ['apple','guava','cherry']
list(filter(lambda x:x.startswith('a'),fruits))
```

reduce

```python
# sum of all item
import functools
l = [1,2,3,4,5]
functools.reduce(lambda x,y:x+y,l)
```

```python
# find min
functools.reduce(lambda x,y:x if x<y else y,[23,11,45,10,1])
```

Enumerate function in python

```python
l = ['Bappy', 'Alex','kumar','ahmed']
for index,name in enumerate(l): # for indexing purpose of elements
    print(index,name)
```

```python
l = ['Bappy', 'Alex','kumar','ahmed']
for index,name in enumerate(l):
    if index == 3:
        print(name)
```
