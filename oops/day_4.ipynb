How object access attributes in OOP

```python
class Person:
    def __init__(self,name,country):
        self.name = name        # attributes == calss variable
        self.country = country
    
    def greeting(self):
        if self.country == "india":
            print(f"Namaste {self.name}")
        else:
            print(f"Hello {self.name}")
```

```python
prn = Person("Anish", "india")
prn.greeting()
```

```python
prn.name
```

```python
prn.country
```

```python
# Attribute can be created from outside of the class
prn.gender = "male"
```

```python
prn.gender
```

```python
prn.age = 23
```

```python
prn.age
```
