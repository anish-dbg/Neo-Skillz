```python
class Shape:
    def area(self,radius):
        return 3.14*radius**2

    def area(self,l,b):
        return l*b

s = Shape()
s.area(3)
print(s)
```

```python
class Shape:
    def area(self,radius,l,b):
        if b == None:
            return 3.14*radius**2
        else:
            return l*b


s = Shape()
result = s.area(radius =4,l = None,b =None)
print(result)
```

Abstraction:
- Abstraction is a fundamental concept in python programming that allows us to simplify complex concepts and focus on the essential details.
- It involves hiding unneccessary details and exposing only the relevant
  information to the users.

```python
# Abstract Base Class ( ABC)
from abc import ABC, abstractmethod

#senior
class BankServer(ABC):
    def database(self):
        print("Connnected to database")
    def login(self):
        print("Connnected to login")

    @abstractmethod
    def security(self):
        print("Security passed")

class MobileApp(BankServer):
    def mobile_login(self):
        print("Mobile app login succcessful")

    def security(self):
        print("Mobile app security")
```

```python
mob = MobileApp()
```

```python
mob.database()
```

```python
mob.login()
```

```python
from abc import ABC, abstractmethod

# 1. THE CONTRACT / ABSTRACT CLASS
class WeddingInvitation(ABC):
    
    @abstractmethod
    def get_venue(self):
        """CONSTRAINT: Junior devs MUST implement this!"""
        pass

    @abstractmethod
    def get_date(self):
        """CONSTRAINT: Junior devs MUST implement this!"""
        pass

    # ABSTRACTION: Hiding the crazy background preparation from the guest
    def attend_wedding(self):
        self._manage_bargaining()
        self._setup_mandap()
        self._cook_food()
        print(f"🎉 Guest successfully arrived at {self.get_venue()} on {self.get_date()}!")

    def _manage_bargaining(self):
        # Hidden complex logic
        pass

    def _setup_mandap(self):
        # Hidden complex logic
        pass

    def _cook_food(self):
        # Hidden complex logic
        pass


# 2. JUNIOR DEV (Brother/Sister) IMPLEMENTATION
class MyFamilyWedding(WeddingInvitation):
    
    def get_date(self):
        return "December 25th"

    # OOP CONSTRAINTS IN ACTION:
    # If your siblings forget to write 'def get_venue(self):',
    # Python throws a TypeError immediately and blocks the card creation!

    def get_venue(self):
        return "Taj Hotel, Mumbai"


# 3. GUEST USAGE
invitation = MyFamilyWedding()
invitation.attend_wedding()
```

Imagine you are managing a wedding. Behind the scenes, there is immense complexity—decorating the mandap, shopping for ingredients, bargaining with the halwai, and coordinating caterers.

However, for a guest, all that complex background logic is hidden; they only need a simple invitation card to attend.

In this setup, your brother and sister act as junior developers tasked with designing the card. But if they forget to include critical details like the venue address, the guest receives an incomplete invitation and cannot attend—causing the system to fail.
