```python
# super keyword :  in child class we can access parent class attribute and method with use use of super keyword
  # method overriding
class Phone:
    def __init__(self, price, brand, camera):
        print("Inside Phone Constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    
    # getter and setter
    def get_price(self):
        return self.__price

    def buy(self):
        print("Buying a Phone")

class SmartPhone(Phone):
    def buy(self):
        print("Buying a Phone inside Smarthphone constructor")

        super().buy()
       # super().brand

s = SmartPhone(20000,"Apple",3)
s.buy()
```

```python
class Phone:
    def __init__(self,price,brand,camera):
        print("Inside Phone constructor")
        self.price = price
        self.brand = brand
        self.camera = camera

class SmartPhone(Phone):
    def __init__(self,price,brand,camera,os,ram):
        print("Inside smartphone constructor")
        super().__init__(price,brand,camera)
        self.os = os
        self.ram = ram

s = SmartPhone(20000,"samsung",3,'android',16)
print(s.brand)
print(s.os)
print(s.price)
```

#Types of inheritance in OOP
- Single Inheritence
- Multilevel Inheritence
- Hierachical Inheritence
- Multiple Inheritence

# multilevel Inheritance

```python
#grandfather

class Product:
    def review(self):
        print("Product customer review")

# Father
class Phone(Product):
    def __init__(self,price,brand,camera):
        print("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print("Buying a phone")

# child
class SmartPhone(Phone):
    pass


s = SmartPhone(20000, "samsung",3)
s.buy()
s.review()
```

# Hierachical Inheritance

```python
class Phone:
    def __init__(self,price,brand,camera):
        print("Inside Phone constructor")
        self.price = price
        self.brand = brand
        self.camera = camera


    def buy(self):
        print("Buying a phone")

class SmartPhone(Phone):
    pass

class FeaturePhone(Phone):
    pass

S = SmartPhone(10000,"apple",3)
S.buy()
```

```python
F = FeaturePhone(10000,"apple",3)
F.buy()
```

# Multiple Inheritance


```python
#parent
class Phone:
    def __init__(self,price,brand,camera):
        print("Inside Phone constructor")
        self.price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print("Buying a phone")

#parent 
class Product:
    def __init__(self):
        pass

    def review(self):
        print("Product customer review")


# child class
class SmartPhone(Phone, Product):
    pass

s = SmartPhone(10000, "apple",3)
s.buy()
s.review()
```
