Aggregation ( Has a relationship)
- one class owns other class

```python
class Customer:
    def __init__(self,name,address,gender):
        self.name = name
        self.gender = gender
        self.address = address

    def print_info(self):
        print(f"Name: {self.name}, Gender: {self.gender}, Address:{self.address.city},{self.address.pin},{self.address.state}")

class Address:
    def __init__(self,city,pin,state):
        self.city = city
        self.pin = pin
        self.state = state
```

```python
adds = Address("india",12345,"new delhi")
cus = Customer("Anish",adds,"male")
```

```python
cus.print_info()
```
