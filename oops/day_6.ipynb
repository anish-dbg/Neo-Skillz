what is instance variable in OOP?
- a variable that can store different data in the same variable for the different object

```python
class Person:
    def __init__(self, name, country):
        self.name = name
        self.country = country
```

```python
p1 = Person("Bappy", "UK")
```

```python
p2 = Person("Alex","USA")
```

```python
p3 = Person("Anish","Canada")
```

```python
print(id(p1),id(p2),id(p3))
```
