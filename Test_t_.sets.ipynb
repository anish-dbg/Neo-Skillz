## Sets
A set is an unordered collection of items. Every set element is unique (no duplicates) and must be immutable (cannot be changed).
however, a set itself is mutable. we can add or remove items from it.

sets can also be used to perform mathematical set operations like union, intersection, symmetric difference, etc

characteristics:
  - Unodered
  - Mutable
  - No Duplicates
  - Can't contain mutable data types

```python
# Creating sets
s = {}
print(type(s))
```

```python
# empty set
s = set()
print(s)
```

```python
# 1d 
s1 = {1,2,3}
print(s1)
```

```python
#2D
s2 = {1,2,3,{4,5}} # sets ke ander sets nhi kr sakte hai
print(s2)
```

```python
s2= {1,2,3,(4,5)}  ## tuples are immutable
print(s2)
```

```python
# homo and hetro
s3 = {1,'hello',4.5,(1,2,3)}
print(s3)
```

```python
# using type conversion
s4 = set([1,2,3])
print(s4)
```

```python
# duplicates not allowed
s5 = {1,1,2,2,3,3}
print(s5)
```

```python
# set can't have mutable items
s6 = {1,2,[3,4]}  # sets ke ander hum mutable items nhi rkh sakte hai
print(s6)
```

```python
# unordered collection of items
s1 = {1,2,3}
s2 = {3,2,1}
s1 == s2
```

```python
# Accessing Items
s1 = {1,2,3}
s1[0]
```

```python
# slicing bhi nhi kr skte hai
s1[0:1]
```

```python
# Editing items
s1 = {1,2,3}
s1[0] = 100
```

```python
# adding items
s = {1,2,3,4}
s.add(5) # we can add only one item
print(s)
```

```python
s.update([11,12,13])
print(s)  # we can add multiple item
```

```python
del s[0]
```

```python
# deleting items
s = {1,2,3,4,5}
s.discard(5) # agr koi element nhi bhi hoga na toa koi error nhi hoga
print(s)
```

```python
# remove
s = {1,2,3,4,5,50}
s.remove(500)  # agr nhi hoga toa error aayega
print(s)
```

```python
# pop
s = {1,2,3,4,5,50}
s.pop()    # randomly delete one element
print(s)
```

```python
# clear
s = {1,2,3,4,5,50}
s.clear()
print(s)
```

```python
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}
```

```python
# Union(|)

s1 | s2    # dono ko mila ke unique value dega
```

```python
# intersection(&)
s1 & s2     # dono mai se jo common element hot hai usko return krta hai
```

```python
# difference(-)
s1-s2    # s1 mai se s2 ko hta do
```

```python
s2 - s1
```

```python
# symmetric difference -  unique elements
s2 ^ s1
```

```python
# membership operator

s1 = {1,2,3,4,5}
1 not in s1
```

```python
# iteration
for i in s1:
    print(i)
```
