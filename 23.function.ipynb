# args and kwargs
- this are special python keywords that are used to pass the variable length of arguments to a function

args
- allows us to pass a variable number of non-keyword arguments to a function

```python
def multiply(a,b,c,d):
    return a*b*c*d
```

```python
multiply(2,3,4,5)
```

```python
def multiply(*args):
    print(args)
```

```python
multiply(1,2,3,4,5,6)
```

```python
def multiply(*args):
    product = 1
    for i in args:
        product = product * i
    return product
```

```python
res = multiply(1,2,3,4,5,6,7,8,8)
print(res)
```

# kwargs
- kwargs allows us to pass any number of keyword arguments

keyword arguments mean that they contain a key-value pair,like a python dictionary

```python
def display(**kwargs):
    print(kwargs)
    for (key,value) in kwargs.items():
        print(f"{key} -> {value}")
```

```python
display(india="Delhi", Bangladesh="Dhaka", srilanka="colombo")
# dictionary
```

```python
# how functions are executed in memory
def odd_even(num): # 1 execute
    if num % 2 == 0:  # 3 execute
        return "Even Number"
    else:
        return "Odd Number"
print(odd_even(5)) # 2 execute

#  a temporary functional block in memory created where the execution held when the execution is complete then it automatically remove from memory
```

Variable Scope in Functions
- Global Variables
- Local Variables

```python
def g(y):
    print(x)
    print(x+1)
x = 5   # global variable : we can access global variable anywhere in the program, we can't change global variable
g(x)
print(x)

# local variable means within function 
```

```python
def h(y):
    x += 1
x = 5
h(x) # we can't change global variable
print(x)
```

```python
def h(y):
    global x
    x += 1
x = 5
h(x)
print(x)
```

Nested Function
- function inside function means it's a loop of function

```python
def f():
    def g():
        print("Inside function g")
    g() # insided function mai call krna hai
    print("Inside function f")
f()
```

```python
g() # we can't access local function outside
```

```python
def f():
    def g():
        print("Inside function g")
        f()  # maximum recursion error
    g() # insided function mai call krna hai
    print("Inside function f")
f()
```
